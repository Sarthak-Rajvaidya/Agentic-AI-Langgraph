First we will understand what is reducer function though :-
In LangGraph, a reducer function tells the graph how to combine a new value with the existing value in the state when multiple nodes update the same state field.


The reducer decides:

"How should I combine the old value and the new value?"

For example, if you want to append instead of overwrite:

Old: [A, B]
New: [C]

Reducer → [A, B, C]

Without a reducer, the new value may replace the old value:

Old: [A, B]
New: [C]

Result → [C]
