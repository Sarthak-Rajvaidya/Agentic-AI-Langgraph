```python
from langgraph.graph import StateGraph,START,END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv

```

```python
load_dotenv()
model = ChatOpenAI()
```

```python
#define state

class BlogState(TypedDict):
    title:str
    outline:str
    content:str
    
```

```python
#create_outline function

def create_outline(create:BlogState)->BlogState:
    
    
    #fetch title
    title = state['title']
    
    #call llm gen outline
    prompt = f'Generatea  detailed outline for a blog on the topic-{title}'
    outline = model.invoke(prompt).content
    #update state
    state['outline'] = outline
    
    return state
```

```python
#create_blog function

def create_blog(state:BlogState)->BlogState:
    
    title = state['title']
    outline = state['outline']
    
    prompt = f'Write a detailed blog on teh title -{title} using the following outline\n {outline}'
    
    content = model.invoke(prompt).content
    
    state['content'] = content
    
    return state
```

```python
#creating graph

graph = StateGraph(BlogState)

#nodes

graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)
```
