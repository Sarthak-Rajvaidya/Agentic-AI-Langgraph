#LAter on discussing we are using example of batamsn whwere he has scored run by balls and 4's aand 6's and we need to calculate strike rate , bounbdary percenatage and balls by boundary 


```python
from langgraph.graph import StateGraph,START,END

from typing import TypedDict
```

```python
class BatsmanState(TypedDict):
    runs:int
    
    
    balls : int
    fours : int
    sixes : int
    
    #To calculate
    sr:float
    bpb:float
    boundary_percentage:float
    summary : str
    
    
```

```python
#strike rate calculation fn

def calculate_sr(state:BatsmanState):
    sr = (state['runs']/state['balls'])/100
    
    
    return {'sr':sr}
```

```python
#calculate balls per boundary fn

def calculate_bpb(state:BatsmanState):
    bpb = state['balls']/(state['fours']+state['sixes'])
    
    
    
    return {'bpb':bpb}
```

```python
def calculate_boundary_percentage(state:BatsmanState):
    boundary_percentage = (
        (state['fours'] * 4 + state['sixes'] * 6)
        / state['runs']
    ) * 100

    return {
        'boundary_percentage': boundary_percentage
    }
```

```python
#summary state

def summary(state:BatsmanState):
    summary = f"""
    Strike Rate - {state['sr']}\n
    Balls per boundary - {state['bpb']}\n
    
    Boundary percent -{state['boundary_percent']}
    
    """
    
    
    
    
    return {'summary':summary}
    
```

```python
graph = StateGraph(BatsmanState)

# Nodes
graph.add_node("calculate_sr", calculate_sr)
graph.add_node("calculate_bpb", calculate_bpb)
graph.add_node("calculate_boundary_percentage", calculate_boundary_percentage)
graph.add_node("summary", summary)

# Edges
graph.add_edge(START, "calculate_sr")
graph.add_edge(START, "calculate_bpb")
graph.add_edge(START, "calculate_boundary_percentage")

graph.add_edge("calculate_sr", "summary")
graph.add_edge("calculate_bpb", "summary")
graph.add_edge("calculate_boundary_percentage", "summary")

graph.add_edge("summary", END)

# Compile
workflow = graph.compile()
```

```python
intial_state = {
    'runs':100,
    'balls':50,
    'fours':6,
    'sixes':4
}

workflow.invoke(intial_state)
```
