Here we will build a basic chatbot uisng Langgaph where we will message and then we will store its message history as well

```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict,Literal,Annotated
from langchain_core.messages import BaseMessage,HumanMessage
from langchain_groq import ChatGroq
```

```python
#state  Reducer function - to maintain history as wehere we chat
from langgraph.graph import add_messages
class ChatState(TypedDict):
    messages:Annotated[list[BaseMessage],add_messages]
```

But what actually is reducer function :-?
Reducer function defines how langgraph should combine a node's new state update with the existing state value

Your chatbot needs to retain the previous messages so the LLM can understand the conversation's context.

Without a reducer, a new update to messages normally replaces the existing list.

```python
llm =  ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

def chat_node(state:ChatState):
    #take user query from state
    messages = state['messages']
    
    #send to llm
    response = llm.invoke(messages)
    #response store state
    
    return {'messages':[response]}
```

```python
graph = StateGraph(ChatState)

#add nodes

graph.add_node('chat_node',chat_node)

graph.add_edge(START,'chat_node')
graph.add_edge('chat_node',END)
```
